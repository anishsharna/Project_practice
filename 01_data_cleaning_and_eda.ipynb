```python
Q1 = df['Rent'].quantile(0.25)
print(Q1)
Q3 = df['Rent'].quantile(0.75)
print(Q3)
IQR = Q3 - Q1
print(IQR)
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print(lower, upper)
```

```python
df_clean = df[(df['Rent'] >= lower) & (df['Rent'] <= upper)]
print(df.shape, df_clean.shape)
```

```python
df_clean.groupby('City')['Rent'].agg(['mean', 'median', 'count'])
```

```python
df_clean = df_clean.copy()
print(df_clean)
```

```python
df_clean['rent_per_sqft'] = df_clean['Rent'] / df_clean['Size']
df_clean.groupby('City')['rent_per_sqft'].median().sort_values(ascending =False)
```

```python
df2 = df[df['Size'] >= 100].copy()
```

```python
q1 = df2.groupby('City')['Rent'].transform(lambda x: x.quantile(0.25))
q3 = df2.groupby('City')['Rent'].transform(lambda x: x.quantile(0.75))

iqr = q3 - q1

mask = (df2['Rent'] >= q1 - 1.5*iqr) & (df2['Rent'] <= q3 + 1.5*iqr)
df_clean = df2[mask].copy()

print(df.shape, df_clean.shape)
```

```python
print(df['City'].value_counts() - df_clean['City'].value_counts())
```
